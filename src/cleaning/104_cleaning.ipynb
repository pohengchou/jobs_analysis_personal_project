# 薪資欄位處理
<p>薪資欄位: </p>
<p>1.「薪資下限」、「薪資上限」兩欄</p>
<p>2.【月薪】【時薪】【年薪】混雜</p>
<p>3.需多為0的資料</p>

```python
df['薪資是否為面議']= np.where(df['薪資下限']==0,1,0) #增加'薪資是否為面議'欄位
df['薪資下限']=df['薪資下限'].replace(0,40000)    #將面議的改成法規規定最低40000
df=df[df['薪資下限']>=1000]     #將時薪(非正職、學生實習)的資料去除

#自動判斷月薪超過300000為年薪，並除以12，轉為月薪
df['薪資下限'] = df['薪資下限'].astype(float)   #原本欄位是(int)類型，除12後會變float，會相衝，故先轉成float。
df.loc[df['薪資下限'] > 300000, '薪資下限']/= 12
df['薪資下限'] = df['薪資下限'].round(0).astype(int)
```

### 產生對照字典，並依據最小的值去對照(學歷最低要求水準)(*非本次分析重點，可略過)

```python
edu_map = {
            1: '不拘',
            2: '高中',
            3: '專科',
            4: '大學',
            5: '碩士',
            6: '博士',
        }
df['學歷要求']=df['學歷要求'].apply(lambda x:edu_map.get(min(ast.literal_eval(x))))
```

### 透過公司人數，增加「公司規模」欄位，判斷是【大型公司】【中型公司】【小型公司】(*非本次分析重點，可略過)

```python
df['員工人數'] = df['員工人數'].apply(lambda x: int(x.replace('人', '')) if isinstance(x, str) and x.endswith('人') else pd.NA) # 去'人'轉數字，並將'暫不提供'改為NA
```

```python
#增加[公司規模]欄位，依員工人數劃分
def classify_company_size(num):       
    if pd.isnull(num):
        return '未知'
    elif num<50:
        return '小型'
    elif num < 200:
        return '中型'
    else:
        return '大型'
df['公司規模']=df['員工人數'].apply(classify_company_size)
```

# 處理「條件要求」欄位
<p>該欄資料是一串dictionary，可以拆成許多欄位。</p>
<p>本次處理重點為，拆出「專長工具」與「工作經驗」這兩個欄位。</p>
<p>「專長工具」可以做為後續【核心技能要求】的分析。</p>

```python
import pandas as pd
import ast # 確保 ast 模組已被導入
import re  # 導入正則表達式模組，用於更精確的關鍵字匹配

# --- 請您自行擴充和完善這個工具關鍵字列表 ---
# 這個列表越完整，從文本中提取的效果越好。
# 建議將常用、特定領域的工具名稱都加進去。
TOOL_KEYWORDS = [
    'Python', 'R', 'SQL', 'Excel', 'Tableau', 'PowerBI', 'Hadoop', 'Spark', 'Kafka',
    'TensorFlow', 'Pytorch', 'Keras', 'Scikit-learn', 'NumPy', 'Pandas', 'Matplotlib', 'Seaborn',
    'AWS', 'Azure', 'GCP', 'Docker', 'Kubernetes', 'Git', 'Linux', 'Java', 'C++', 'Scala', 'Go',
    'SAS', 'SPSS', 'ETL', 'API', 'SSIS', 'SSRS', 'SSAS', 'Power BI', 'QlikView', 'D3.js', 'NoSQL',
    'MongoDB', 'PostgreSQL', 'MySQL', 'Redshift', 'BigQuery', 'Snowflake', 'Airflow', 'Luigi',
    'Jupyter', 'VS Code', 'GitLab', 'GitHub', 'CI/CD', 'Machine Learning', 'Deep Learning',
    '統計', '數據分析', '數據挖掘', '機器學習', '深度學習', '資料庫', '雲端', '自動化',
    'BI', 'Dashboard', '報表', '可視化', '自然語言處理', 'NLP', '計算機視覺', 'CV',
    'A/B Testing', '模型部署', 'MLOps', 'Data Bricks', 'Databricks', 'Dataproc',
    'Looker Studio', 'Snowflake', 'GCP' # <--- 已將這些新的工具關鍵字加入
]
# 將關鍵字按長度降序排列，以避免短詞匹配先於長詞（例如先匹配 'SQL' 再匹配 'MySQL'）
TOOL_KEYWORDS = sorted(TOOL_KEYWORDS, key=len, reverse=True)


def parse_string(x):
    """
    將字串轉換為 Python 的資料結構（如字典、列表、數字等）。
    """
    if isinstance(x, str):
        try:
            return ast.literal_eval(x)
        except Exception as e:
            # 您可以選擇在這裡加入錯誤處理或日誌記錄
            # print(f"Error parsing: {x} - {e}")
            return {} # 解析失敗時返回空字典，方便後續處理
    return x

def extract_all_tools_from_row(row):
    """
    從 '條件要求' 字典內的 'specialty' 和 '其他條件'，以及獨立的 'description' 欄位中，
    提取所有相關的工具名稱。此函數需要對 DataFrame 進行 row-wise 應用 (df.apply(..., axis=1))。
    """
    extracted_tools = set() # 使用集合來自動處理重複項

    # 輔助函數：用於在文本中更精確地查找關鍵字（使用詞邊界）
    def find_keywords_in_text(text, keywords_list):
        if not isinstance(text, str):
            return []
        found = set()
        text_lower = text.lower()
        for keyword in keywords_list:
            # 使用正則表達式，並加上 \b 詞邊界，確保只匹配完整的詞
            # re.escape 用於處理關鍵字中可能存在的正則表達式特殊字元 (例如 C++, D3.js)
            if re.search(r'\b' + re.escape(keyword.lower()) + r'\b', text_lower):
                found.add(keyword)
        return list(found)

    # --- 1. 從 '條件要求' 字典的 'specialty' 提取 ---
    job_require_dict = row.get('條件要求') # 安全地獲取 '條件要求' 欄位的值
    if isinstance(job_require_dict, dict) and 'specialty' in job_require_dict:
        specialty_list = job_require_dict['specialty']
        if isinstance(specialty_list, list):
            for s in specialty_list:
                if isinstance(s, dict) and s.get('description'):
                    desc_from_specialty = s['description']
                    # 將 specialty 的 description 直接加入（如果它本身就是一個工具名）
                    if isinstance(desc_from_specialty, str):
                        extracted_tools.add(desc_from_specialty)
                    
                    # 同時檢查 specialty 的 description 是否包含預定義的工具關鍵字
                    found_in_desc = find_keywords_in_text(desc_from_specialty, TOOL_KEYWORDS)
                    extracted_tools.update(found_in_desc)

    # --- 2. 從 '條件要求' 字典的 '其他條件' 提取 ---
    if isinstance(job_require_dict, dict) and '其他條件' in job_require_dict:
        other_conditions_text = job_require_dict['其他條件']
        found_in_other = find_keywords_in_text(other_conditions_text, TOOL_KEYWORDS)
        extracted_tools.update(found_in_other)

    # --- 3. 從獨立的 'description' 欄位提取 ---
    description_text = row.get('description') # 從行中安全地獲取獨立的 'description' 欄位值
    found_in_main_desc = find_keywords_in_text(description_text, TOOL_KEYWORDS)
    extracted_tools.update(found_in_main_desc)
    
    # 將提取到的所有工具名稱合併為一個逗號分隔的字串，並去除重複項
    if extracted_tools:
        # 排除空字符串或None，並排序以保持輸出一致性
        final_tools = [tool for tool in extracted_tools if tool and pd.notna(tool)]
        return ", ".join(sorted(final_tools))
    else:
        return None

# --- 以下是應用程式碼 ---

# 假設 df DataFrame 已經被載入
# df = pd.read_csv('your_data.csv') # 如果還沒載入，請在這裡載入

# 1. 將 '條件要求' 字串轉換為 Python 的資料結構
df['條件要求'] = df['條件要求'].apply(parse_string)

# 2. 提取工作經驗 (保持與您原有程式碼一致)
df['工作經驗'] = df['條件要求'].apply(lambda x:x.get('workExp') if isinstance(x,dict) else None)

# 3. 提取所有工具名稱 (取代原本的 '專長工具' 提取邏輯)
# 注意：這裡使用 df.apply(..., axis=1) 因為 extract_all_tools_from_row 需要訪問整行數據
df['所有工具'] = df.apply(extract_all_tools_from_row, axis=1)

# 4. 刪除原始的 '條件要求' 欄位 (保持與您原有程式碼一致)
df.drop(columns='條件要求', inplace=True)

# --- 完成處理，您可以檢查新生成的欄位 ---
# print(df[['工作經驗', '所有工具']].head())
```
